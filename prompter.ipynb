```python
import os
import json

token_comuni = "./json_files/new_tokenizer_tests/common_tokens.json"

with open(token_comuni, 'r', encoding='utf-8') as file:
    data = json.load(file)

topics = set()
    
for key, value in data.items():
    for a in value['topic']:
        topics.add(a)

for topic in topics:
    print(topic)
```

```python
for t in topics:
    tokens = []
    for key, value in data.items():
        if t in value['topic']:
            tokens.append(key)
    with open(f'./json_files/topic_list/{t}.txt', 'w', encoding='utf-8') as file:
        file.write('\n'.join(tokens))
    
```

```python
import random

def walk_directory(directory):
    l = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(".txt"):
                l.append(os.path.join(root, file))
    return l

for file in walk_directory('./json_files/topic_list'):
    with open(file, 'r', encoding='utf-8') as f:
        content = f.read()
    chosen_words = random.sample(content.splitlines(), 10)
    print(f'\"{os.path.basename(file).split(".")[0]}\":{chosen_words}')
```
