```python
import json
import os
import random

def walk_directory(directory):
    l = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if not file.endswith(".csv"):
                l.append(os.path.join(root, file))
    #print(f"Found {len(l)} files in {directory}")
    return l

# for all keys in selected_words.json, copy the corresponding file from the id provided and paste it into a new directory
def copy_selected_files(selected_words_file, source_directory, destination_directory):
    with open(selected_words_file, 'r') as f:
        selected_words = json.load(f)

    if not os.path.exists(destination_directory):
        os.makedirs(destination_directory)

    for key in selected_words:
        file_id = selected_words[key]['id']
        sane_key = key.replace(" ", "_").replace("/", "_").lower()
        in_dir_path = f'{source_directory}/{sane_key}/{file_id}.json'
        with open(in_dir_path, 'r', encoding='utf-8') as src_file:
            content = json.load(src_file)
        out_file_name = f'{sane_key}_{file_id}.json'
        out_dir_path = f'{destination_directory}/{out_file_name}'
        with open(out_dir_path, 'w', encoding='utf-8') as dest_file:
            json.dump(content, dest_file, indent=4)
        

if __name__ == "__main__":
    selected_words_file = './json_files/topic_list/selected_words.json'
    source_directory = './distribution_divided'
    destination_directory = './selected_files_output'
    copy_selected_files(selected_words_file, source_directory, destination_directory)
```
